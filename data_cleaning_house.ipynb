```python
df['unit_price'] = df['unit_price'].str.replace(' ', '').str.extract('(\d+)')[0]
```

```python
df['unit_price'] = pd.to_numeric(df['unit_price'], errors='coerce')
```

```python
df['Sahə'] = df['Sahə'].str.extract('(\d+)')[0].astype(float)
```

```python
mask = df['unit_price'].isna()
df.loc[mask, 'unit_price'] = df.loc[mask, 'price'] / df.loc[mask, 'Sahə']
```

```python
df['Mərtəbə']
```

```python
df['Mərtəbə'] = df['Mərtəbə'].str.extract('(\d+)')[0]
```

```python
df['Mərtəbə'] = df['Mərtəbə'].astype(float)
```

```python
median_value = df[df['Kateqoriya'].isin(['Köhnə tikili', 'Yeni tikili'])]['Mərtəbə'].median()

def fill_floor(row):
    if pd.isna(row['Mərtəbə']):
        if row['Kateqoriya'] in ['Köhnə tikili', 'Yeni tikili']:
            return median_value
        else:
            return 0
    return row['Mərtəbə']

df['Mərtəbə'] = df.apply(fill_floor, axis=1)

df.head()
```
